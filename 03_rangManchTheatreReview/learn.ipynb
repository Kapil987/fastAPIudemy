```python
class Review:
    def __init__(self, play_name, reviewer_name, rating, comment):
        self.play_name = play_name
        self.reviewer_name = reviewer_name
        self.rating = rating
        self.comment = comment
```

```python
db_review = Review(
    play_name="Hamlet",
    reviewer_name="John",
    rating=5,
    comment="Excellent performance"
)
```

```python
review_data = {
    "play_name": "Hamlet",
    "reviewer_name": "John",
    "rating": 5,
    "comment": "Excellent performance"
}
```

```python
db_review = Review(**review_data)
```
